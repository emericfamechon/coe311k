```python
import numpy as np
import math
import random
import matplotlib.pyplot as plt
```

```python
def taylor_calc(terms):
    therange = np.arange(0,10,0.1)
    for x in therange:
        result = 0
    for n in range(terms):
        factorial = math.factorial(2*n+1)
        numerator = ((-1)**n)*(x**(2*n+1))
        taylor = numerator/factorial
        result = result + taylor
    error = np.abs(result - np.sin(x))
    return error

```

```python
term1 = taylor_calc(1)
term2 = taylor_calc(2)
term3 = taylor_calc(3)
term4 = taylor_calc(4)

print(term1,"\n",term2,"\n",term3,"\n",term4,"\n")
results = [term1, term2, term3, term4]
```

```python
domain = np.arange(0,4)

plt.subplot(2, 2, 1) 
plt.plot(domain, results, 'r--')

```

### Discuss

Beginning at x=1, the error is a number around 10. As it increases towards x=5, the error increases up to around 1300 at x=5. From there, it begins to decerase rapidly and by x=12, the error is already far below 1. As a result, the error does not decrease uniformly, but rather increases to a peak before finally descending towards 0. This is demonstrated in the graph below where I plotted some higher x-values in comparison to the x=1, 2, 3, 4 I had previously graphed.
