# Jacobi

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def jacobi(A, b, x0, tol, max_iterations):
    errors=[]
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        ierror=np.linalg.norm(x_new-x, ord=np.inf)
        errors.append(ierror)
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1, errors
        x = x_new
        print(x)
    return x, max_iterat, errors
```

### Tolerance 10e-3

```python
A = np.array([[3, -0.1, -0.2], [0.1, 7, -0.3], [0.3, -0.2, 10]])
b = np.array([7.85, -19.3, 71.4])
x0 = np.zeros_like(b)
tol = 1e-3
max_iterations = 100
solution, iterations, jerrors = jacobi(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
je3=jerrors
```

# Gauss-Seidel

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    errors=[]
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) 
            # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) 
            # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
            ierror=np.linalg.norm(x_new-x, ord=np.inf)
            errors.append(ierror)
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1, errors
        x = x_new
        print(x)
    return x, max_iterations, errors
```

### Tolerance 10e-3

```python
A = np.array([[3, -0.1, -0.2], [0.1, 7, -0.3], [0.3, -0.2, 10]])
b = np.array([7.85, -19.3, 71.4])
x0 = np.zeros_like(b)
tol = 1e-3
max_iterations = 100
#Print/save current time
solution, iterations, gerrors = gauss_seidel(A, b, x0, tol, max_iterations)
#Print/save current time
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
ge3=gerrors
```

# Discuss

As the tolerance becomes smaller, the number of iterations required increases. This is true for both Gauss-Seidel and Jacobi. In bother cases, as the tolerance exponent becomes more and more negative, the iterations go up. Jacobi goes from 4 to 7 to 9 while Gauss-Seidel goes 4 to 5 to 7.
