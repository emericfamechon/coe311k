```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
h = 0.001  # step size
x_start, x_end = 0, 0.05
num_steps = int((x_end - x_start) / h)
x_values = np.linspace(x_start, x_end, num_steps + 1)
y_values = np.zeros(num_steps + 1)

# Initial condition
y_values[0] = 0  # y(0) = 0
# Function representing dy/dx = -2y
def f(x, y):
    return -1000*y+3000-2000*np.exp(-x)

# Implement Euler's method
for i in range(num_steps):
    y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    
    
#Analytical Solution
def analytic_s(x):
    return 3-0.998*np.exp(-1000*x)-2.002*np.exp(-x)

y_analytic = analytic_s(x_values)

# Plot the results
plt.plot(x_values, y_values, label="Euler's Method Approximation")
plt.plot(x_values, y_analytic, label="Analytical Solution")
plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler vs Analytical")
plt.legend()
plt.show()
```
