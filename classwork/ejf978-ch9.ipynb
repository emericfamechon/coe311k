## Part 1

```python
import numpy as np

# Data points 

X = np.array([1, 2, 3, 4, 5])

Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)

A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X, Y)

x2=np.linspace(0,6,100)

y2=0.9500000000000001*x2+0.5900000000000001

plt.plot(x2,y2)

plt.show()
```

## Part 2

```python
import random
rng = np.random.default_rng()

# Data points 

X_rand = rng.uniform(0,10,size=5)
X_rand.sort()
Y_rand = rng.uniform(0,10,size=5)

# Stack the design matrix with a column of ones (for the intercept)

A = np.vstack([X_rand, np.ones(len(X_rand))]).T

# Solve using least squares

m, c = np.linalg.lstsq(A, Y_rand, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
```

```python
plt.scatter(X_rand, Y_rand)

x2=np.linspace(0,10,100)

y2=-m*x2+c

plt.plot(x2,y2)

plt.show()
```

### The fit is much poorer than the previous attempt form Part 1. The points are completely scattered and there is no veritable semblance between the line and the points.
