```python
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
rng = np.random.default_rng()

a1=rng.random(size=(3,3))
a2=rng.random(size=(3,3))

b1=rng.random(size=(6,6))
b2=rng.random(size=(6,6))

c1=rng.random(size=(12,12))
c2=rng.random(size=(12,12))

d1=rng.random(size=(24,24))
d2=rng.random(size=(24,24))

e1=rng.random(size=(48,48))
e2=rng.random(size=(48,48))

f1=rng.random(size=(96,96))
f2=rng.random(size=(96,96))

g1=rng.random(size=(192,192))
g2=rng.random(size=(192,192))

h1=rng.random(size=(384,384))
h2=rng.random(size=(384,384))
```

```python
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
```

```python
m1 = [a1,b1,c1,d1,e1,f1,g1,h1]
m2 = [a2,b2,c2,d2,e2,f2,g2,h2]
```

```python
times_naive = []

for i in range(len(m1)):
    start_naive = time.perf_counter()
    naive_matrix_multiplication(m1[i], m2[i])
    end_naive = time.perf_counter()
    time_naive = end_naive - start_naive
    times_naive.append(time_naive)
    
print(times_naive)
```

```python
times_numpy = []

for i in range(len(m1)):
    start_numpy = time.perf_counter()
    np.dot(m1[i],m2[i])
    end_numpy = time.perf_counter()
    time_numpy = end_numpy - start_numpy
    times_numpy.append(time_numpy)
    
print(times_numpy)
```

```python
sizes = ["3x3","6x6","12x12","24x24","48x48","96x96","192x192","384x384"]

plt.plot(sizes, times_naive,'r--',label="Naive")
plt.plot(sizes,times_numpy,'b--',label="Numpy")
plt.title("Numpy vs Naive Run Times")
plt.ylabel("Run Time (seconds)")
plt.xlabel("Matrix Sizes")
plt.legend()
plt.show()
```

As evidenced by the graph and the printed arrays, we can determine that for the lower matrix sizes, the naive version runs faster than the numpy one. However, As the matrix size increases, the naive run time drastically increases while the numpy run time stays fairly fast and close to 0. This means that for smaller matrices, it would be more efficient to use the naive version while for larger matrices (in this case 48x48 and up), the numpy method is way more efficient.
