## Part a

```python
import time
```

```python
import numpy as np
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start_time = time.perf_counter()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            end_time = time.perf_counter()
            return x_new, k, (end_time - start_time)
        x = x_new
        print(x)
    end_time = time.perf_counter()
    return x, max_iterations, (end_time - start_time)
```

```python
# Example usage
A = np.array([[10,-1,2,0,0],[-1,11,-1,3,0],[2,-1,10,-1,0],[0,3,-1,8,-2],[0,0,0,-2,9]])
b = np.array([14,30,26,25,37])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
solution, iterations, elapsed_time = jacobi(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {elapsed_time:.8f}")
```

```python
import numpy as np
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start_time = time.perf_counter()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) 
            # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) 
            # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            end_time = time.perf_counter()
            return x_new, k, (end_time - start_time)
        x = x_new
        print(x)
    end_time = time.perf_counter()
    return x, max_iterations, (end_time - start_time)
```

```python
# Example usage
A = np.array([[10,-1,2,0,0],[-1,11,-1,3,0],[2,-1,10,-1,0],[0,3,-1,8,-2],[0,0,0,-2,9]])
b = np.array([14,30,26,25,37])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
solution, iterations, elapsed_time = gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {elapsed_time:.8f}")
```

## DISCUSSION

For part a, both of the methods converge. However, the Jacobi iteration runs one iteration while the Gauss-Seidel requires two iterations. In addition to this, the Jacobi iteration required approximately .0003 less time to run than the Gauss-Seidel. 

For part b, neither of the methods converged as they both reached the maximum runs (100) without having found a true solution. They both took around .035 seconds although the Jacobi was still faster than the Gauss-Seidel by a fraction of a second.
