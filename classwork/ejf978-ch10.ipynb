```python
import numpy as np
import matplotlib.pyplot as plt
```

# PART A

```python
np.random.seed(0)
x = np.linspace(-5, 5, 50)
y = 2 * x**2 + 3 * x + np.random.randn(50) * 10
```

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

def r_squared(y_true, y_pred):
    ss_total = np.sum((y_true - np.mean(y_true))**2)
    ss_res = np.sum((y_true - y_pred)**2)
    return 1 - (ss_res/ss_total)
```

```python
A = np.vstack([x, np.ones(len(x))]).T
m,c = np.linalg.lstsq(A,y,rcond=None)[0]
y_ls = m*x + c
mse_ls = mse(y,y_ls)
r2_ls = r_squared(y,y_ls)
print("MSE:", mse_ls)
print("R^2:", r2_ls)
```

```python
degree = 3

coeffs = np.polyfit(x,y,degree)
poly = np.poly1d(coeffs)

y_poly = poly(x)

mse_poly = mse(y,y_poly)
r2_poly = r_squared(y,y_poly)

print("MSE:", mse_poly)
print("R^2:", r2_poly)
```

## PART A DISCUSSION

The polynomial fit is definitely the better fit due to the higher r-squared value as well as the much lower MSE value. The 372 MSE and .05 R^2 for the lstsqr shows that it is clearly not a good fit for the data points. On the other hand, the poly has a R^2 value of .728 which is in the perfect range to depict a good fit. 

# PART B

```python
for deg in range(1,37):

    coeffs = np.polyfit(x,y,deg)
    poly = np.poly1d(coeffs)

    y_pred = poly(x)

    m = mse(y,y_pred)
    r2 = r_squared(y,y_pred)

    print(f"Degree {deg}: MSE={m:.3f}, R^2={r2:.3f}")
```

Here we can see that the polynomial begins to blatantly overfit at around degree 36. It is most likely earlier than this, however it is at this point that we begin to recieve errors. On the other side, degree 1 seems to be the most liklely candidate for the underfit. It is at this point that the R^2 value drops immensely and the MSE shoots up.

# PART C

```python
degree = 3

V = np.vander(x, degree+1)

coeffs_v = np.linalg.lstsq(V,y,rcond=None)[0]

y_vander = V @ coeffs_v

mse_v = mse(y,y_vander)
r2_v = r_squared(y,y_vander)

print("\nVandermonde Polynomial Fit")
print("MSE:", mse_v)
print("R^2:", r2_v)
```
